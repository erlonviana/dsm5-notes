Gerando 100.000 linhas para testarmos 

```python
import random
import numpy as np
import importlib

perceptron_module = importlib.import_module("01_perceptron_numpy")
PerceptronNumPy = perceptron_module.PerceptronNumPy

#Configuração de consistencia (para o random sempre adicionar o numero abaixo)
random.seed(42)
np.random.seed(42)

#Gerar 100000 amostras (dias hospedados, gasto total)
amostras_list = [[ random.uniform(1,15), random.uniform(10,500)] for _ in range(100000)]
amostras_numpy = np.array(amostras_list)

# Parametros do perceptron
pesos_init = [0.8, 0.3]
bias_init = -7.0

modelo_np = PerceptronNumPy(pesos=pesos_init, bias=bias_init)
```

```python
# For em Python Puro
def predict_pure_python(entradas, pesos, bias):
    z = bias
    for i in range(len(entradas)):
        z += entradas[i] * pesos[i]
    return 1 if z >= 0 else 0
```

# Função %%timeit para o notebook mostrar quanto tempo esta execução demorou

```python
%%timeit
resultados_puro = []
for c in amostras_list:
    res = predict_pure_python(c, pesos_init, bias_init)
    resultados_puro.append(res)
```

```python
%%timeit
resultados_numpy = modelo_np.predict(amostras_numpy)
```
