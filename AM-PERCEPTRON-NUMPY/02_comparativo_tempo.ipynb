Este script serve para preparar um teste de desempenho (benchmark) do tempo de execução do seu PerceptronNumPy

Gerando 100.000 linhas para testarmos 

```python
#Importa o módulo random da biblioteca padrão do Python
import random
#Importa a biblioteca NumPy com o alias np
import numpy as np
#Importa o módulo importlib da biblioteca padrão.
import importlib

#Carrega o arquivo 01_perceptron_numpy.py como um módulo Python
perceptron_module = importlib.import_module("01_perceptron_numpy")
#Atribui a classe a uma variável local
PerceptronNumPy = perceptron_module.PerceptronNumPy

#Configuração de consistencia (para o random sempre adicionar o numero abaixo)
#Garante que os números aleatórios gerados sejam sempre os mesmos toda vez que o código rodar
#assim é possível a reprddutibilidade do teste
## Semente para o módulo random padrão
random.seed(42)
# Semente para o NumPy
np.random.seed(42)

#Gerar 100.000 amostras (dias hospedados, gasto total)
# Para cada uma das 100.000 iterações:
#   - Gera um número aleatório entre 1 e 15 (dias)
#   - Gera um número aleatório entre 10 e 500 (gasto)
#   - Cria uma lista com [dias, gasto]
#   - Adiciona à lista amostras_list
amostras_list = [[ random.uniform(1,15), random.uniform(10,500)] for _ in range(100000)]
#Transforma a lista Python em um array NumPy 2D Formato: 100.000 linhas × 2 colunas
amostras_numpy = np.array(amostras_list)

# Parametros do perceptron
pesos_init = [0.8, 0.3]
bias_init = -7.0

#Instanciando o modelo
modelo_np = PerceptronNumPy(pesos=pesos_init, bias=bias_init)
```

Este código mostra a implementação manual (em Python puro) da função de predição do Perceptron, sem usar NumPy

```python
# For em Python Puro
def predict_pure_python(entradas, pesos, bias):
    z = bias
    for i in range(len(entradas)):
        z += entradas[i] * pesos[i]
    return 1 if z >= 0 else 0
```

Função %%timeit para o notebook mostrar quanto tempo esta execução demorou

Este código está medindo o tempo de execução da versão em Python puro usando o %%timeit do Jupyter Notebook

%%timeit é um magic command do IPython/Jupyter Notebook que Executa o código várias vezes para medir o tempo médio

```python
%%timeit
resultados_puro = []
for c in amostras_list:
    res = predict_pure_python(c, pesos_init, bias_init)
    resultados_puro.append(res)
```

Mede o tempo que a versão vetorizada com NumPy leva para classificar as 100.000 amostras

```python
%%timeit
resultados_numpy = modelo_np.predict(amostras_numpy)
```
